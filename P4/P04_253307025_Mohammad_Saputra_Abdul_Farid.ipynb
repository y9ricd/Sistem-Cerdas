### <font color="red">Kegiatan 2</font>

```python
def bfs(graf, awal, tujuan):
    antrian = [[awal]]
    while antrian:
        jalur = antrian.pop(0)
        kota = jalur[-1]
        if kota == tujuan:
            return jalur
        for tetangga in graf[kota]:
            if tetangga not in jalur:
                antrian.append(jalur + [tetangga])
    return None

rute = bfs(peta, "Madiun", "Surabaya")
print(" -> ".join(rute))
```

### <font color="red">Penjelasan singkat kegiatan 2</font>
- Algoritma BFS mencari rute dari Madiun ke Surabaya dengan prioritas jumlah langkah/titik tersedikit.

- Menggunakan struktur data antrian (queue) berkonsep FIFO (First-In, First-Out), di mana lintasan terdepan diambil menggunakan antrian.pop(0).

- Pengecekan if tetangga not in jalur dilakukan untuk mencegah traversal berputar (infinite loop) pada graf.

### <font color="red">Adaptasi Kegiatan 2</font>

```python
# Adaptasi: BFS mencari jalur obby dengan jumlah rintangan paling sedikit
def bfs_obby(graf, awal, tujuan):
    antrian = [[awal]]                        # antrian berisi JALUR, bukan titik
    while antrian:
        jalur = antrian.pop(0)                # FIFO: ambil jalur PALING DEPAN
        titik = jalur[-1]                     # titik terakhir pada jalur itu
        if titik == tujuan:
            return jalur                      # sampai! kembalikan jalurnya
        for tetangga in graf[titik]:
            if tetangga not in jalur:          # jangan mutar balik
                antrian.append(jalur + [tetangga])
    return None                                # antrian habis = tak ada jalur

rute_obby = bfs_obby(peta_obby, "Start", "Finish")
print(" -> ".join(rute_obby))
```

### <font color="red">Kegiatan 3</font>

```python
def dfs(graf, awal, tujuan):
    tumpukan = [[awal]]
    while tumpukan:
        jalur = tumpukan.pop()
        kota = jalur[-1]
        if kota == tujuan:
            return jalur
        for tetangga in graf[kota]:
            if tetangga not in jalur:
                tumpukan.append(jalur + [tetangga])
    return None

print(" -> ".join(dfs(peta, "Madiun", "Surabaya")))
```

### <font color="red">Penjelasan singkat Kegiatan 3</font>  
- Algoritma DFS menelusuri graf dengan prioritas pencarian mendalam pada satu jalur hingga mentok sebelum mencoba rute alternatif.

- Menggunakan struktur data tumpukan (stack) berkonsep LIFO (Last-In, First-Out), di mana jalur paling akhir ditambahkan akan diambil terlebih dahulu menggunakan tumpukan.pop().

### <font color="red">Adaptasi Kegiatan 3</font>

```python
# Adaptasi Kegiatan 3: DFS menelusuri jalur obby sampai mentok sebelum mencoba jalur lain
def dfs_obby(graf, awal, tujuan):
    tumpukan = [[awal]]
    while tumpukan:
        jalur = tumpukan.pop()                # LIFO: ambil jalur PALING BELAKANG
        titik = jalur[-1]
        if titik == tujuan:
            return jalur
        for tetangga in graf[titik]:
            if tetangga not in jalur:
                tumpukan.append(jalur + [tetangga])
    return None

rute_obby_dfs = dfs_obby(peta_obby, "Start", "Finish")
print(" -> ".join(rute_obby_dfs))
```

### <font color="red">Kegiatan 4</font>

```python
jarak = {
    "Madiun":    {"Ngawi": 32, "Ponorogo": 29, "Nganjuk": 47},
    "Ngawi":     {"Madiun": 32},
    "Ponorogo":  {"Madiun": 29},
    "Nganjuk":   {"Madiun": 47, "Kediri": 28, "Jombang": 40},
    "Kediri":    {"Nganjuk": 28, "Jombang": 39, "Malang": 79},
    "Jombang":   {"Nganjuk": 40, "Kediri": 39, "Mojokerto": 24},
    "Mojokerto": {"Jombang": 24, "Surabaya": 45, "Malang": 61},
    "Malang":    {"Kediri": 79, "Mojokerto": 61, "Surabaya": 94},
    "Surabaya":  {"Mojokerto": 45, "Malang": 94},
}
h = {"Madiun": 140, "Ngawi": 155, "Ponorogo": 160, "Nganjuk": 100,
     "Kediri": 95, "Jombang": 60, "Mojokerto": 40, "Malang": 65, "Surabaya": 0}

def a_star(jarak, h, awal, tujuan):
    terbuka = [(h[awal], 0, [awal])]
    while terbuka:
        terbuka.sort()
        f, g, jalur = terbuka.pop(0)
        kota = jalur[-1]
        if kota == tujuan:
            return jalur, g
        for tetangga, km in jarak[kota].items():
            if tetangga not in jalur:
                g_baru = g + km
                f_baru = g_baru + h[tetangga]
                terbuka.append((f_baru, g_baru, jalur + [tetangga]))
    return None, None

rute, total = a_star(jarak, h, "Madiun", "Surabaya")
print(" -> ".join(rute), f"| total {total} km")
```

### <font color="red">Penjelasan singkat Kegiatan 4</font>
- Algoritma A* (A-Star) menelusuri graf menggunakan fungsi evaluasi $f(n) = g(n) + h(n)$ untuk memprioritaskan pencarian pada rute dengan estimasi total biaya terkecil.
- Menggabungkan biaya jarak tempuh aktual ($g$) dan perkiraan jarak heuristik ($h$) ke titik tujuan agar pencarian rute terpendek lebih efisien dibandingkan BFS maupun DFS.

### <font color="red">Adaptasi Kegiatan 4</font>

```python
# Adaptasi Kegiatan 4: A* menghitung jalur tercepat menuju Finish (jarak)
jarak_obby = {
    "Start":             {"Platform_A": 5, "Platform_B": 6},
    "Platform_A":        {"Start": 5, "Rintangan_Goyang": 4, "Checkpoint_1": 8},
    "Platform_B":        {"Start": 6, "Rintangan_Api": 5},
    "Rintangan_Goyang":  {"Platform_A": 4, "Checkpoint_1": 3},
    "Rintangan_Api":     {"Platform_B": 5, "Checkpoint_1": 6},
    "Checkpoint_1":      {"Platform_A": 8, "Rintangan_Goyang": 3, "Rintangan_Api": 6, "Platform_C": 7},
    "Platform_C":        {"Checkpoint_1": 7, "Rintangan_Lompat": 4, "Finish": 9},
    "Rintangan_Lompat":  {"Platform_C": 4, "Finish": 3},
    "Finish":            {"Platform_C": 9, "Rintangan_Lompat": 3},
}
# heuristik: perkiraan jarak lurus menuju Finish
h_obby = {"Start": 15, "Platform_A": 11, "Platform_B": 12, "Rintangan_Goyang": 8,
          "Rintangan_Api": 9, "Checkpoint_1": 7, "Platform_C": 4, "Rintangan_Lompat": 2, "Finish": 0}

def a_star_obby(jarak, h, awal, tujuan):
    terbuka = [(h[awal], 0, [awal])]           # (f, g, jalur)
    while terbuka:
        terbuka.sort()                          # f terkecil ke depan
        f, g, jalur = terbuka.pop(0)            # ambil yang paling menjanjikan
        titik = jalur[-1]
        if titik == tujuan:
            return jalur, g                     # jalur + total jarak
        for tetangga, jrk in jarak[titik].items():
            if tetangga not in jalur:
                g_baru = g + jrk                # jarak nyata sejauh ini
                f_baru = g_baru + h[tetangga]   # + perkiraan sisa
                terbuka.append((f_baru, g_baru, jalur + [tetangga]))
    return None, None

rute_terpendek, total_jarak = a_star_obby(jarak_obby, h_obby, "Start", "Finish")
print(" -> ".join(rute_terpendek), f"| total {total_jarak} unit")
```

BFS cukup dipakai ketika semua langkah dianggap setara dan yang dibutuhkan hanyalah rute dengan jumlah langkah paling sedikit, tanpa peduli jarak sebenarnya - misalnya untuk estimasi cepat saat semua ruas jalan dianggap sama. DFS jarang dipakai untuk pencarian rute karena sifatnya menyelam satu jalur sampai mentok tanpa mempertimbangkan efisiensi sama sekali - bisa saja DFS menghasilkan rute yang jauh lebih panjang, bergantung urutan tetangga dalam dictionary, sehingga tidak cocok dipakai pada kondisi yang menuntut kecepatan dan optimalitas seperti pencarian rute tercepat.
