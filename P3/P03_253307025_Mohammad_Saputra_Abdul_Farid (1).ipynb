1. **Dua ekspresi disebut ekuivalen jika nilainya sama pada semua baris tabel kebenaran**

    | p | q | p ∧ q | ¬(p ∧ q) | ¬p | ¬q | ¬p ∨ ¬q |
    | --- | --- | --- | --- | --- | --- | --- |
    | B | B | B | **S** | S | S | **S** |
    | B | S | S | **B** | S | B | **B** |
    | S | B | S | **B** | B | S | **B** |
    | S | S | S | **B** | B | B | **B** |

    Kolom ¬(p ∧ q) dan ¬p ∨ ¬q memiliki nilai yang sama pada keempat baris (S, B, B, B). Jadi terbukti bahwa ¬(p ∧ q) ekuivalen dengan ¬p ∨ ¬q.

2. **a) Semua praktikum dikumpulkan melalui LMS**

    logika predikatnya: P(x) = x adalah praktikum; L(x) = x dikumpulkan melalui LMS

        ∀x (P(x) → L(x))

    Artinya, untuk semua x, jika x ialah praktikum maka x dikumpulkan lewat LMS.

    **b) Ada mahasiswa yang belum mengumpulkan P05**

    logika predikat: M(x) = x adalah mahasiswa; K(x) = x mengumpulkan P05

        ∃x (M(x) ∧ ¬K(x))

    Artinya, ada x yang merupakan mahasiswa dan x tidak mengumpulkan P05.

    **c) Setiap mahasiswa yang mengumpulkan tepat waktu mendapat nilai penuh**

    logika predikat: M(x) = x adalah mahasiswa; T(x) = x mengumpulkan tepat waktu; N(x) = x mendapat nilai penuh

        ∀x ((M(x) ∧ T(x)) → N(x))

    Artinya, untuk semua x, jika x mahasiswa dan x mengumpulkan tepat waktu, maka x mendapat nilai penuh.

3. **Lima aturan produksi (domain: troubleshooting WiFi kos)**

    R1: JIKA lampu_indikator_mati
    MAKA masalah_listrik

    R2: JIKA masalah_listrik DAN kabel_power_longgar
    MAKA cek_colokan

    R3: JIKA lampu_indikator_menyala DAN tidak_ada_sinyal
    MAKA masalah_pemancar

    R4: JIKA masalah_pemancar DAN router_terlalu_panas
    MAKA restart_router

    R5: JIKA restart_router DAN sinyal_tetap_hilang
    MAKA hubungi_provider

    Contoh penalaran berantai: kasus lampu indikator menyala tapi tidak ada sinyal, dan ternyata router terlalu panas. R3 terpicu sehingga "masalah_pemancar" menjadi fakta baru. Fakta ini menjadi kondisi untuk R4, sehingga R4 terpicu dan menghasilkan fakta baru "restart_router". Kalau setelah direstart sinyal tetap hilang, R5 ikut terpicu dan kesimpulannya adalah hubungi_provider. Kesimpulan satu aturan menjadi pemicu aturan berikutnya — seperti itu rantai penalarannya.

4. **Relasi antar simpul:**

    *Farid -is-a-> Mahasiswa*

    *Mahasiswa -belajar-> Sistem Cerdas*

    *Sistem Cerdas -diajar_oleh-> Pak Fajar*

    *Mahasiswa -anggota-> Prodi TI*

    *Prodi TI -berada_di-> Jurusan Teknik*

    (6 simpul: Farid, Mahasiswa, Sistem Cerdas (MK), Pak Fajar (Dosen), Prodi TI (Prodi), Jurusan Teknik (Gedung))

    Contoh pewarisan: karena Farid is-a Mahasiswa, dan Mahasiswa anggota Prodi TI, maka Farid otomatis ikut mewarisi sifat "anggota Prodi TI" tanpa perlu dibuatkan panah khusus dari Farid ke Prodi TI.

5. **Kasus penguin**

    Penguin is-a burung, tetapi penguin tidak bisa terbang. Solusinya, frame penguin menimpa (override) slot yang diwariskan dari frame burung:

    FRAME: Burung        |         FRAME: Penguin

    adalah  : Hewan       |         adalah  : Burung

    sayap   : ada         |         sayap   : (diwarisi: ada)

    terbang : bisa   ---->|         terbang : tidak   <- ditimpa               |         renang  : bisa   <- khas penguin

    Berdasarkan perbandingan di atas, penguin tetap mewarisi sifat umum burung (memiliki sayap). Namun, khusus pada slot kemampuan "terbang", nilainya sengaja ditimpa menjadi "tidak".

    Kalau di dunia nyata, kaidah yang menyatakan "semua X pasti memiliki sifat Y" hampir selalu punya pengecualian. Karena itu, sebuah sistem representasi pengetahuan tidak cukup hanya mengandalkan fitur pewarisan sifat (inheritance), tapi juga harus mendukung mekanisme penimpaan nilai (overriding). Inilah salah satu akar masalah kenapa sistem pakar sering membengkak dan sulit dirawat ketika basis aturannya sudah mencapai ribuan, karena setiap pengecualian kasus harus ditangani satu per satu secara manual.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/y9ricd/Sistem-Cerdas/main/P3/retail_sales_dataset.csv"
retail_df = pd.read_csv(url)

print(retail_df.head())
print(retail_df.info())
print(retail_df.describe())

# kolom baru: kategori transaksi besar/kecil berdasarkan Total Amount
retail_df["kategori_transaksi"] = retail_df["Total Amount"].apply(
    lambda x: "Besar" if x > 500 else "Sedang" if x > 100 else "Kecil"
)

transaksi_besar = retail_df[retail_df["kategori_transaksi"] == "Besar"]
print(transaksi_besar[["Customer ID", "Product Category", "Total Amount"]])

print(retail_df.groupby("Product Category")["Total Amount"].mean())
print(retail_df.groupby("Gender")["Total Amount"].sum())
```

**1. NumPy - Vektorisasi**
Vektorisasi adalah kemampuan untuk melakukan perhitungan pada seluruh kumpulan angka sekaligus, tanpa harus melakukan perulangan (looping) secara manual seperti pada list biasa. Konsep ini penting karena AI pada umumnya menangani data dalam jumlah sangat besar (jutaan angka), sehingga pendekatan "per-kolom" jauh lebih efisien dibanding pendekatan "per-elemen".

**2. Pandas - "Excel-nya Python"**
Pandas digambarkan sebagai "Excel-nya Python" karena menangani data dalam bentuk tabel yang disebut DataFrame, lengkap dengan label kolomnya sendiri. Kemampuan yang dipelajari pada library ini meliputi: membaca file CSV, menghitung statistik data, menyaring baris berdasarkan kriteria tertentu, menambahkan kolom baru, dan mengelompokkan data.

**3. Matplotlib - Memilih Grafik yang Tepat**
Matplotlib digunakan untuk memilih jenis tampilan grafik yang tepat guna menjawab pertanyaan tertentu: histogram untuk melihat distribusi satu variabel, grafik batang untuk perbandingan antar kategori, grafik garis untuk melihat tren, dan scatter plot (plot titik) untuk melihat hubungan antara dua variabel.
