### <font color="red">Latihan 2</font>

````python
def konversi_nilai(angka):
    if angka > 82:
        return "A"
    elif angka > 75:
        return "AB"
    elif angka > 67:
        return "B"
    elif angka > 59:
        return "BC"
    elif angka > 49:
        return "C"
    elif angka > 39:
        return "D"
    else:
        return "E"

print("1. Nilai 82 = ",konversi_nilai(82))
print("2. Nilai 75 = ",konversi_nilai(75))
print("3. Nilai 67 = ",konversi_nilai(67))
````

###<font color="red">penjelasan singkat latihan 2
**(def)** : membungkus kode ke dalam sebuah Fungsi, jika memakai analogi (def) itu seperti sebuah mesin blender yang bisa mengubah buah(input) menjadi jus(output/return)  
**(if / elif / else)** : kontrol untuk meng eksekusi kode tertentu berdasarkan kondisi berdasarkan pemenuhan kriteria atau suatu kondisi logis,Jika suatu kondisi bernilai benar (True), blok kode di dalamnya dijalankan dan sisa kondisi di bawahnya diabaikan.

**if** angka > 82: mengecek apakah nilai lebih besar dari 82.

**elif** angka > 75: mengecek kondisi berikutnya jika rentang pertama tidak terpenuhi.

**else**: berfungsi sebagai penampung untuk kondisi terakhir (nilai di bawah ambang batas sebelumnya).

**Mencetak mahasiswa di atas rata-rata beserta huruf mutu**: Loop kedua ini mengecek satu per satu: apakah mhs["nilai"] lebih besar dari rata_kelas?
