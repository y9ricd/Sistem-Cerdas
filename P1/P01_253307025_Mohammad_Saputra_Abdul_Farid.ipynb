JAWABAN

1. **Menalar**: sistem menarik kesimpulan dari pengetahuan yang sudah direpresentasikan (aturan/logika). Contoh: sistem pakar diagnosa penyakit yang menyimpulkan diagnosis dari gejala lewat aturan IF–THEN yang diberikan dokter.

    **Belajar**: sistem menemukan pola sendiri dari data, tanpa aturan ditulis manual. Contoh: sistem rekomendasi e-commerce yang mempelajari pola belanja dan percarian jutaan pengguna untuk menyarankan produk.

    **Beradaptasi**: sistem menyesuaikan perilakunya ketika situasi/lingkungan berubah. Contoh: mobil yang mengubah kecepatan dan rute saat kondisi jalan tiba-tiba berubah (hujan, macet, ada pejalan kaki).  

  2. Lompatan besar AI terjadi saat tiga syarat bertemu sekaligus: algoritma, data, dan daya komputasi. Backpropagation (algoritmanya) memang sudah ditemukan 1986, tapi saat itu belum ada dua syarat lain: data berlabel berskala besar (seperti ImageNet, yang baru tersedia lewat internet) dan daya komputasi murah yang sanggup melatih jaringan berlapis-lapis (GPU baru terjangkau sekitar 2012).

  3. **Performance**: menu yang direkomendasikan benar-benar dipilih/disukai mahasiswa, kepuasan mahasiswa, peningkatan variasi menu terjual, kecepatan sistem menampilkan rekomendasi.  
     **Environment**: mahasiswa dengan selera dan riwayat beli berbeda-beda, daftar menu harian kantin, jam makan siang yang ramai, stok bahan yang terbatas.  
     **Actuators**: menampilkan daftar menu rekomendasi di layar/aplikasi kantin, mengirim notifikasi promo menu tertentu.  
     **Sensors**: riwayat pemesanan mahasiswa, rating/ulasan menu, hari dan jam saat itu, data stok menu yang tersedia.

  4. **Ojek daring**: rumpun statistik/pembelajaran memprediksi permintaan & waktu tempuh dari data historis, rumpun simbolik (algoritma pencarian) menentukan rute tercepat di peta, dan rumpun evolusioner bisa dipakai mengoptimalkan penempatan armada driver terhadap banyak permintaan sekaligus.

  5. Era sistem pakar (MYCIN, XCON) justru gagal karena kebanyakan aturan bikin biaya perawatan pengetahuan membengkak, memicu AI winter kedua. Machine learning lahir sebagai jawabannya: kecerdasan diukur dari kemampuan menangani situasi yang tak sepenuhnya terprogram, bukan dari jumlah aturan.

**Penggunaan Tanda Pagar (#):** Baris yang diawali dengan simbol ini murni berfungsi sebagai catatan pribadi untuk programmer. Komputer akan mengabaikannya dan tidak akan mengeksekusinya sebagai program.  

**Perintah print():** Fungsi ini ibarat perintah untuk "tampilkan pesan ini ke layar". Karena pesan yang ingin dimunculkan berupa kalimat ("Halo, saya akan belajar Sistem Cerdas"),("Nama saya: Farid") maka teksnya wajib diapit menggunakan tanda kutip.

```python
# Kegiatan 4
# Python bisa langsung dipakai berhitung
jp_teori = 3 # jumlah JP teori
jp_praktik = 5 # jumlah JP praktik
menit_per_jp = 50 # 1 JP = 50 menit
total_menit = (jp_teori + jp_praktik) * menit_per_jp
print("Total menit kuliah hari ini:", total_menit)
```

###<font color="red">Penjelasan
**Pembuatan Variabel:** Baris seperti jp_teori = 3 adalah cara kita membuat wadah data. Simbol sama dengan (=) di sini bertugas untuk menyuntikkan angka 3 ke dalam wadah bernama jp_teori.  

**Operasi Hitung:** Python bisa bertindak layaknya kalkulator matematis. Pada kode (jp_teori + jp_praktik) * menit_per_jp, Python akan menjumlahkan angka di dalam kurung terlebih dahulu, lalu mengalikannya (menggunakan simbol bintang *) dengan angka 50.  

**Menggabungkan Teks dan Variabel:** Pada perintah print baris terakhir, tanda koma (,) digunakan sebagai jembatan penghubung. Ini memungkinkan layar menampilkan teks kalimat biasa sekaligus angka hasil hitungan akhir secara berdampingan.
